```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([1,2,3,4,5])
y = np.array([0,0,0,1,1])
```

```python
X
```

```python
y
```

```python
b0 = 0
b1 = 0
alpha = 0.1
iterations = 1000
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
for i in range(iterations):

    z = b0 + b1 * X
    y_pred = sigmoid(z)

    error = y_pred - y

    db1 = np.mean(error * X)
    db0 = np.mean(error)

    b1 = b1 - alpha * db1
    b0 = b0 - alpha * db0

print("Intercept:", b0)
print("Coefficient:", b1)
```

```python
def predict(x):
    z = b0 + b1 * x
    p = sigmoid(z)
    return 1 if p >= 0.5 else 0

print("Prediction for 1 hours:", predict(1))
```

```python
x_vals = np.linspace(0,10,100)

z_vals = b0 + b1 * x_vals
sigmoid_vals = sigmoid(z_vals)

plt.figure(figsize=(8,5))

# Plot sigmoid curve
plt.plot(x_vals, sigmoid_vals, label="Sigmoid Curve", color="blue")

# Plot original data
plt.scatter(X, y, color="red", label="Training Data")

plt.xlabel("Study Hours")
plt.ylabel("Probability / Class")
plt.title("Logistic Regression with Sigmoid Curve")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
```

```python
X = np.array([[1],[2],[3],[4],[5]])
y = np.array([0,0,0,1,1])
```

```python
model = LogisticRegression()
```

```python
model.fit(X,y)
```

```python
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
hours = [[1]]
prediction = model.predict(hours)
probability = model.predict_proba(hours)
```

```python
print("Prediction:", prediction)
print("Probability:", probability)
```
